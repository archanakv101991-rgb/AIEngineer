# User Login System

This notebook demonstrates a simple user login system implementation.

```python
import hashlib
import getpass

# Simple user database (in production, use a real database)
users_db = {
    'admin': hashlib.sha256('password123'.encode()).hexdigest(),
    'user': hashlib.sha256('user456'.encode()).hexdigest()
}

print("User Database initialized with sample users")
```

```python
def hash_password(password):
    """Hash a password using SHA256"""
    return hashlib.sha256(password.encode()).hexdigest()

def login(username, password):
    """Authenticate user with username and password"""
    if username in users_db:
        if users_db[username] == hash_password(password):
            return True, "Login successful!"
        else:
            return False, "Invalid password"
    else:
        return False, "User not found"

def register(username, password):
    """Register a new user"""
    if username in users_db:
        return False, "Username already exists"
    else:
        users_db[username] = hash_password(password)
        return True, "User registered successfully!"

print("Login functions defined")
```

```python
# Test login with correct credentials
success, message = login('admin', 'password123')
print(f"Test 1 - Login with correct password: {message}")

# Test login with incorrect credentials
success, message = login('admin', 'wrongpassword')
print(f"Test 2 - Login with incorrect password: {message}")

# Test login with non-existent user
success, message = login('nonexistent', 'password')
print(f"Test 3 - Login with non-existent user: {message}")
```

```python
# Test registration
success, message = register('newuser', 'newpass789')
print(f"Test 4 - Register new user: {message}")

# Verify the new user can login
success, message = login('newuser', 'newpass789')
print(f"Test 5 - Login with newly registered user: {message}")
```
